```python
# load the encoded representations from synthetic/res/lorenz_stochastic_infonce_sparse/encoded_representations_seed22.pkl
import pickle

# Specify the path to the encoded representations file
# seed 22 refers to the DCA initialization
encoded_repr_path_seed22 = "res/lorenz_stochastic_infonce_sparse/encoded_representations_seed22.pkl"
# seed 20 refers to the random initialization
encoded_repr_path_seed20 = "res/lorenz_stochastic_infonce_sparse/encoded_representations_seed20.pkl"

def load_encoded_representations(encoded_repr_path):
    # Load the encoded representations
    with open(encoded_repr_path, "rb") as f:
        encoded_data = pickle.load(f)

    # The encoded_data may be a dict or list depending on the saving function
    print(f"Loaded encoded representations from {encoded_repr_path}")
    print(f"Type: {type(encoded_data)}")
    if isinstance(encoded_data, dict):
        print(f"Keys: {encoded_data.keys()}")
    elif isinstance(encoded_data, list):
        print(f"Length: {len(encoded_data)}")
    else:
        print(encoded_data)

    # Display the shape of the encoded data if possible
    if isinstance(encoded_data, dict):
        for k, v in encoded_data.items():
            print(f"Key: {k}, shape: {getattr(v, 'shape', 'N/A')}")
    elif isinstance(encoded_data, list):
        if len(encoded_data) > 0 and hasattr(encoded_data[0], 'shape'):
            print(f"List[0] shape: {encoded_data[0].shape}")
        else:
            print("List elements do not have a 'shape' attribute or the list is empty.")
    elif hasattr(encoded_data, "shape"):
        print(f"Shape: {encoded_data.shape}")
    else:
        print("Encoded data does not have a 'shape' attribute.")

    encoded_features = encoded_data['encoded_representations'].detach().numpy()
    return encoded_features
```

```python
import matplotlib.pyplot as plt
import numpy as np
import itertools

def plot_encoded_features(arr):
    ndims = arr.shape[1] if arr.ndim > 1 else 1
    if ndims < 3:
        raise ValueError(f"encoded_features must have at least 3 dimensions, got shape {arr.shape}")

    # Generate all unique combinations of 3 dimensions
    dim_combinations = list(itertools.combinations(range(ndims), 3))

    nplots = len(dim_combinations)
    ncols = min(nplots, 3)
    nrows = (nplots + ncols - 1) // ncols

    fig = plt.figure(figsize=(8 * ncols, 6 * nrows))
    for i, dims in enumerate(dim_combinations):
        ax = fig.add_subplot(nrows, ncols, i + 1, projection='3d')
        ax.plot(arr[:, dims[0]], arr[:, dims[1]], arr[:, dims[2]], color='blue', alpha=0.8, lw=1.0)
        ax.set_xlabel(f'Latent Dim {dims[0]}')
        ax.set_ylabel(f'Latent Dim {dims[1]}')
        ax.set_zlabel(f'Latent Dim {dims[2]}')
        ax.set_title(f'3D Trajectory: Dims {dims[0]}, {dims[1]}, {dims[2]}')
    fig.tight_layout()
    plt.show()
```

```python
# plot the encoded features for the seed 22
encoded_features_seed22 = load_encoded_representations(encoded_repr_path_seed22)
plot_encoded_features(encoded_features_seed22)
```

```python
# plot the encoded features for the seed 20
encoded_features_seed20 = load_encoded_representations(encoded_repr_path_seed20)
plot_encoded_features(encoded_features_seed20)
```
